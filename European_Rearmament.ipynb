## 1. Introduction

### 1.1 Background and Motivation

European security has undergone significant changes over the past decade. Russia's annexation of Crimea in 2014, its full-scale invasion of Ukraine in February 2022, and growing uncertainty surrounding the future of European security arrangements have renewed debates about military expenditure, strategic autonomy, and defence cooperation within the European Union (EU).

These developments raise fundamental questions about Europe's defence capabilities and its ability to respond collectively to geopolitical threats.

Although European governments have announced increases in defence budgets and major military procurement initiatives, the extent and distribution of this rearmament require empirical investigation.

Moreover, European countries face different geopolitical circumstances. While some EU member states share borders with Russia or are located near the conflict in Ukraine, others are geographically more distant. These differences may influence national defence priorities and expenditure patterns.

At the same time, increasing military expenditure does not necessarily translate into greater European strategic autonomy. A significant share of military equipment acquired by European countries originates from foreign suppliers, including the United States.

Understanding European rearmament therefore requires examining not only **how much countries spend**, but also **where they are located** and **where their military equipment comes from**.

This project investigates the evolution of European military expenditure, its relationship with geopolitical geography, and the dependence of EU member states on foreign military suppliers.

3. **Eurostat GISCO:** Country boundaries used for mapping and geographical distance calculations.
4. **Eurostat city reference data:** Capital-city locations used to construct an alternative measure of proximity to Russia.

Military expenditure and arms transfers capture different dimensions of defence activity. Expenditure represents financial resources allocated to military activities, while TIV represents the standardized volume of major weapons delivered internationally.

The indicators are therefore analyzed separately and compared through statistical relationships rather than treated as directly equivalent quantities.

Three main indicators are considered:

Military expenditure measured in constant US dollars accounts for estimated changes in prices over time.

Military expenditure as a percentage of GDP measures the proportion of a country's economic output devoted to military spending.

This indicator is particularly useful for comparing countries of different economic sizes.

**3. Military Expenditure (Current USD)**

Military expenditure in current US dollars measures nominal spending at the prices and exchange rates of the corresponding year.

It provides information about the nominal size of military budgets, but is less appropriate for measuring real expenditure growth across time.

Additional socioeconomic and geographical data may be obtained from Eurostat and Eurostat GISCO.

```python

# Load the three relevant datasets
milex_constant = pd.read_excel(
    sipri_file,
    sheet_name="Constant (2024) US$",
    header=5
)

milex_gdp = pd.read_excel(
    sipri_file,
    sheet_name="Share of GDP",
    header=5
)

milex_current = pd.read_excel(
    sipri_file,
    sheet_name="Current US$",
    header=5
)
```

```python

# Display the first rows
display(milex_constant.head(10))

# Inspect the column names
print(milex_constant.columns.tolist())

# Inspect the dataset dimensions
print("Shape:", milex_constant.shape)
```

```python

# Countries to inspect
countries = ["Germany", "France", "Italy", "Poland"]

# Select the relevant years
years = [2014, 2021, 2022, 2025]

# Select countries and columns
sample = milex_constant.loc[
    milex_constant["Country"].isin(countries),
    ["Country"] + years
]

display(sample)
```

```python

# List of the 27 EU member states
eu27 = [
    "Austria", "Belgium", "Bulgaria", "Croatia",
    "Cyprus", "Czechia", "Denmark", "Estonia",
    "Finland", "France", "Germany", "Greece",
    "Hungary", "Ireland", "Italy", "Latvia",
    "Lithuania", "Luxembourg", "Malta", "Netherlands",
    "Poland", "Portugal", "Romania", "Slovakia",
    "Slovenia", "Spain", "Sweden"
]

print("Number of EU countries:", len(eu27))
```

```python

# Check which EU countries are not found in SIPRI
missing_countries = set(eu27) - set(milex_constant["Country"].dropna())

print("Countries not found:", missing_countries)
```

```python

# Define the study period
years = list(range(2014, 2026))

# Filter countries and select relevant columns
eu_milex = milex_constant.loc[
    milex_constant["Country"].isin(eu27),
    ["Country"] + years
].copy()

# Check the result
print("Dataset shape:", eu_milex.shape)

display(eu_milex.head())
```

```python

# Count missing and non-numeric observations by year
numeric_values = eu_milex[years].apply(
    pd.to_numeric,
    errors="coerce"
)

missing_by_year = numeric_values.isna().sum()

print("Missing observations by year:")
display(missing_by_year)
```

```python

# Ensure expenditure values are numeric
eu_milex[years] = eu_milex[years].apply(
    pd.to_numeric, errors="coerce"
)

# Convert from wide to long format
eu_milex_long = eu_milex.melt(
    id_vars="Country",
    value_vars=years,
    var_name="Year",
    value_name="Military_Expenditure"
)

# Ensure Year is an integer
eu_milex_long["Year"] = eu_milex_long["Year"].astype(int)

# Sort by country and year
eu_milex_long = eu_milex_long.sort_values(
    by=["Country", "Year"]
).reset_index(drop=True)

display(eu_milex_long.head(15))
```

```python

print("Dataset shape:", eu_milex_long.shape)

print("Number of countries:",
      eu_milex_long["Country"].nunique())

print("Number of years:",
      eu_milex_long["Year"].nunique())

print("Missing expenditure values:",
      eu_milex_long["Military_Expenditure"].isna().sum())

print("Duplicate country-year observations:",
      eu_milex_long.duplicated(
          subset=["Country", "Year"]
      ).sum())
```

```python

def prepare_milex(df, value_name):
    # Filter EU countries and years
    df = df.loc[
        df["Country"].isin(eu27),
        ["Country"] + years
    ].copy()

    # Convert expenditure values to numeric
    df[years] = df[years].apply(
        pd.to_numeric, errors="coerce"
    )

    # Transform wide to long format
    df = df.melt(
        id_vars="Country",
        value_vars=years,
        var_name="Year",
        value_name=value_name
    )

    df["Year"] = df["Year"].astype(int)

    return df


# Apply the function to both datasets
eu_gdp_long = prepare_milex(
    milex_gdp, "GDP_Share"
)

eu_current_long = prepare_milex(
    milex_current, "Current_Expenditure"
)
```

```python

# Convert GDP share from proportions to percentages
eu_gdp_long["GDP_Share"] = (
    eu_gdp_long["GDP_Share"] * 100
)

# Inspect the results
display(eu_gdp_long.head())
```

```python

# Merge real military expenditure with GDP share
eu_defence = eu_milex_long.merge(
    eu_gdp_long,
    on=["Country", "Year"],
    how="left",
    validate="one_to_one"
)

# Add current military expenditure
eu_defence = eu_defence.merge(
    eu_current_long,
    on=["Country", "Year"],
    how="left",
    validate="one_to_one"
)

# Inspect the result
display(eu_defence.head(15))
```

```python

print("Dataset shape:", eu_defence.shape)

print("\nMissing values:")
print(eu_defence.isna().sum())

print("\nDataset preview:")
display(eu_defence.head())
```

```python

# Calculate total annual EU-27 military expenditure
eu_total = (
    eu_defence.groupby("Year")["Military_Expenditure"]
    .sum(min_count=27)
    .reset_index()
)

# Convert from millions to billions of constant 2024 USD
eu_total["Expenditure_Billion"] = (
    eu_total["Military_Expenditure"] / 1000
)

display(eu_total)
```

```python

# Calculate annual percentage growth
eu_total["Annual_Growth"] = (
    eu_total["Expenditure_Billion"].pct_change(
        fill_method=None
    ) * 100
)

display(eu_total.round(2))
```

```python

fig, ax = plt.subplots(figsize=(11, 6))

# Plot military expenditure
ax.plot(
    eu_total["Year"],
    eu_total["Expenditure_Billion"],
    marker="o",
    linewidth=2
)

# Mark Russia's full-scale invasion of Ukraine
ax.axvline(
    x=2022,
    color="red",
    linestyle="--",
    alpha=0.7,
    label="Russian invasion of Ukraine (2022)"
)

# Labels and title
ax.set_title(
    "EU-27 Military Expenditure (2014–2025)",
    fontsize=15
)

ax.set_xlabel("Year")
ax.set_ylabel("Billion constant 2024 USD")

ax.set_xticks(eu_total["Year"])
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python

# Select expenditure in the first and last years
spending_2014 = eu_total.loc[
    eu_total["Year"] == 2014,
    "Expenditure_Billion"
].iloc[0]

spending_2025 = eu_total.loc[
    eu_total["Year"] == 2025,
    "Expenditure_Billion"
].iloc[0]

# Calculate total percentage growth
total_growth = (
    (spending_2025 / spending_2014) - 1
) * 100

print(f"EU military expenditure in 2014: ${spending_2014:.2f} billion")
print(f"EU military expenditure in 2025: ${spending_2025:.2f} billion")
print(f"Total real growth: {total_growth:.2f}%")
```


### 2.7 Initial Findings: Evolution of EU Military Expenditure

Aggregate military expenditure across the EU-27 increased from approximately **$210.77 billion in 2014 to $430.39 billion in 2025**, measured in constant 2024 US dollars.

This represents a **real increase of 104.19%**, equivalent to approximately $219.62 billion.

The historical trend reveals two important patterns:

1. **Pre-2022 growth:** European military expenditure was already increasing steadily before Russia's full-scale invasion of Ukraine.
2. **Post-2022 acceleration:** Expenditure increased more rapidly in the years following 2022, with particularly strong growth toward the end of the study period.

These results provide strong descriptive evidence of increasing European military expenditure.

However, higher military expenditure does not necessarily correspond to proportional improvements in military capability, and the observed acceleration cannot automatically be attributed entirely to the invasion of Ukraine.


```python

# Separate the annual growth rates into two periods
pre_2022 = eu_total.loc[
    eu_total["Year"].between(2015, 2021),
    "Annual_Growth"
]

post_2022 = eu_total.loc[
    eu_total["Year"].between(2022, 2025),
    "Annual_Growth"
]

# Calculate average annual growth
avg_pre = pre_2022.mean()
avg_post = post_2022.mean()

print(f"Average annual growth (2015–2021): {avg_pre:.2f}%")
print(f"Average annual growth (2022–2025): {avg_post:.2f}%")
print(f"Difference: {avg_post - avg_pre:.2f} percentage points")
```

```python

fig, ax = plt.subplots(figsize=(11, 6))

# Plot annual growth rates
ax.bar(
    eu_total["Year"],
    eu_total["Annual_Growth"]
)

# Separate pre- and post-invasion periods
ax.axvline(
    x=2021.5,
    color="red",
    linestyle="--",
    label="2022: Russian invasion of Ukraine"
)

ax.axhline(y=0, color="black", linewidth=0.8)

ax.set_title(
    "Annual Growth in EU-27 Military Expenditure (2015–2025)",
    fontsize=15
)

ax.set_xlabel("Year")
ax.set_ylabel("Annual real growth (%)")

ax.set_xticks(eu_total["Year"][1:])
ax.legend()
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```


### Findings: Acceleration of European Military Expenditure

The analysis reveals a substantial acceleration in European military expenditure growth after 2022.

Between 2015 and 2021, the EU-27 experienced an average annual real expenditure growth rate of **4.20%**. During 2022–2025, this increased to **11.44%**, representing an acceleration of **7.24 percentage points**.

Three observations emerge:

1. **Rearmament predates 2022.** European military expenditure was already increasing consistently before Russia's full-scale invasion of Ukraine.

2. **Growth accelerated after 2022.** The strongest increases occurred in 2023–2025 rather than immediately in 2022, suggesting a potentially delayed spending response.

3. **The acceleration became progressively stronger.** Annual growth rates increased substantially toward the end of the study period, particularly in 2024 and 2025.

Overall, these results suggest that European rearmament is both a long-term trend and a process that accelerated following the deterioration of the European security environment.

However, the analysis remains descriptive. The observed acceleration cannot be attributed exclusively to Russia's invasion without considering other economic, political, and security factors.


```python

# Select military expenditure in 2014 and 2025
country_comparison = eu_defence[
    eu_defence["Year"].isin([2014, 2025])
].pivot(
    index="Country",
    columns="Year",
    values="Military_Expenditure"
)

# Rename columns
country_comparison = country_comparison.rename(
    columns={
        2014: "Spending_2014",
        2025: "Spending_2025"
    }
)

# Calculate absolute expenditure change (billion USD)
country_comparison["Absolute_Change"] = (
    country_comparison["Spending_2025"]
    - country_comparison["Spending_2014"]
) / 1000

# Calculate percentage growth
country_comparison["Percentage_Growth"] = (
    (country_comparison["Spending_2025"]
     / country_comparison["Spending_2014"] - 1) * 100
)

# Display countries with the largest absolute increases
display(
    country_comparison.sort_values(
        "Absolute_Change",
        ascending=False
    ).head(10).round(2)
)
```

```python

# Rank countries by percentage expenditure growth
display(
    country_comparison.sort_values(
        "Percentage_Growth",
        ascending=False
    ).head(10).round(2)
)
```

Large economies can contribute substantially to aggregate European military expenditure even when their relative growth rates are moderate.

Conversely, smaller countries may register exceptionally high percentage increases while contributing less to the overall EU expenditure increase.

Several geographically exposed countries, particularly Lithuania, Latvia, and Poland, experienced substantial relative growth. This observation provides motivation for the subsequent geospatial analysis.

However, geographical proximity cannot yet be identified as an explanatory factor without systematically examining the relationship across all EU member states.


# 3. Part B — Does Geography Influence European Defence Spending?

## B1. Research Objective and Geographical Methodology

### Research Question

> Is geographical proximity to Russia associated with higher military expenditure and stronger increases in defence spending among EU member states?

Part A demonstrated that European rearmament has been substantial but unevenly distributed across countries.

One potential explanation for these differences is geographical exposure to Russia. Countries located closer to Russian territory may perceive greater security risks and consequently allocate more resources to defence.

This section investigates whether geographical proximity is statistically associated with three indicators:

1. Military expenditure as a percentage of GDP in 2025.
2. Change in military expenditure as a percentage of GDP between 2014 and 2025.
3. Percentage growth in real military expenditure between 2014 and 2025.

### Geographical Methodology

Country boundaries are obtained from Eurostat GISCO and transformed into the ETRS89-extended / LAEA Europe projected coordinate system (EPSG:3035).

Two geographical distance measures are used:

- **Minimum territorial distance:** The shortest straight-line distance between an EU member state's territory and the Russian territory included in the geographical analysis.
- **Capital-city distance:** The shortest straight-line distance between a country's capital and the same Russian geographical reference.

The Russian reference includes Kaliningrad, which is particularly relevant for several EU member states in Central and Eastern Europe.

Countries sharing a land border with Russia are assigned a territorial distance of approximately zero.

These measures provide consistent proxies for geographical proximity, but do not directly measure perceived military threats, strategic vulnerability, or geopolitical relationships.

### Analytical Approach

The analysis combines spatial visualization, Pearson and Spearman correlations, Ordinary Least Squares regression, and sensitivity analyses.

Particular attention is given to whether the findings remain stable when alternative geographical definitions are used or influential observations are excluded.

All relationships are interpreted as observational associations rather than causal effects.

## B4. Regression Analysis and Sensitivity Checks

### Research Objective

The correlation analysis identifies strong negative associations between geographical distance from Russia and several military expenditure indicators.

This section investigates the relationship more formally using Ordinary Least Squares (OLS) regression.

The dependent variable is the change in military expenditure as a percentage of GDP between 2014 and 2025, measured in percentage points.

The main explanatory variable is capital-city distance from Russia, expressed in units of 100 kilometres.

HC3 heteroskedasticity-robust standard errors are used to evaluate uncertainty in the estimated coefficients.

### Additional Checks

The regression analysis is complemented by:

- Residual diagnostics and Cook's Distance.
- Sensitivity tests excluding potentially influential countries.
- Leave-one-out estimation.
- A model including the initial 2014 defence burden as an additional explanatory variable.

The purpose is to assess the stability of the estimated geographical relationship, rather than establish a causal effect of proximity on military expenditure.


```python

import statsmodels.formula.api as smf

# Prepare the regression dataset
reg_data = geo_analysis[
    ["Country", "Capital_Distance_Russia_km", "Change_pp"]
].dropna().copy()

# Express distance in hundreds of kilometres
reg_data["Distance_100km"] = (
    reg_data["Capital_Distance_Russia_km"] / 100
)

# Fit a simple linear regression
model = smf.ols(
    formula="Change_pp ~ Distance_100km",
    data=reg_data
).fit(cov_type="HC3")

# Display regression results
print(model.summary())
```

```python

# Generate predicted values from the regression
x_values = np.linspace(
    reg_data["Distance_100km"].min(),
    reg_data["Distance_100km"].max(),
    100
)

predicted = model.predict(
    pd.DataFrame({"Distance_100km": x_values})
)

# Create scatterplot
fig, ax = plt.subplots(figsize=(11, 7))

ax.scatter(
    reg_data["Distance_100km"] * 100,
    reg_data["Change_pp"],
    alpha=0.75,
    s=65,
    label="EU member states"
)

# Plot fitted regression line
ax.plot(
    x_values * 100,
    predicted,
    color="red",
    linewidth=2,
    label="OLS regression"
)

ax.axhline(0, color="gray", linestyle="--", alpha=0.5)

ax.set_title(
    "Capital Distance to Russia vs Change in Defence Burden",
    fontsize=14
)

ax.set_xlabel("Capital distance to Russia (km)")
ax.set_ylabel("Change in military expenditure (% GDP), 2014–2025 (pp)")

ax.grid(alpha=0.3)
ax.legend()

plt.tight_layout()
plt.show()
```


### Findings: Linear Regression of Defence Burden on Geographical Distance

A simple Ordinary Least Squares (OLS) regression was estimated to investigate the relationship between capital-city distance to Russia and changes in military expenditure as a percentage of GDP between 2014 and 2025.

The model uses 27 EU member states and heteroskedasticity-robust (HC3) standard errors.

**Regression results:**

| Statistic | Result |
|---|---:|
| Distance coefficient (per 100 km) | -0.0762 |
| R-squared | 0.466 |
| Adjusted R-squared | 0.445 |
| Coefficient p-value | 0.002 |
| 95% confidence interval | [-0.123, -0.029] |

The estimated coefficient indicates that an additional 100 km of distance from Russia is associated with an average 0.0762 percentage-point smaller increase in military expenditure burden.

The model accounts for approximately 46.6% of the observed cross-country variation in defence burden changes.

The coefficient is statistically distinguishable from zero under the model assumptions, supporting the existence of a negative linear association between geographical distance and increased defence expenditure burden.

However, the regression does not establish causality. Countries differ in their economic characteristics, historical defence policies, security alliances, and geopolitical circumstances.

Furthermore, the linear specification may not fully capture the relationship, particularly among countries located far from Russia.

Additional diagnostic and robustness analyses are therefore necessary before drawing final conclusions.


```python

# Create a copy of our regression dataset
diagnostics = reg_data.copy()

# Calculate predicted values
diagnostics["Predicted"] = model.fittedvalues

# Calculate residuals
diagnostics["Residual"] = model.resid

# Calculate influence statistics
influence = model.get_influence()

diagnostics["Cooks_D"] = influence.cooks_distance[0]

# Display countries with the highest influence
display(
    diagnostics[
        ["Country", "Change_pp", "Predicted",
         "Residual", "Cooks_D"]
    ]
    .sort_values("Cooks_D", ascending=False)
    .head(10)
    .round(3)
)
```

```python

fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(
    diagnostics["Predicted"],
    diagnostics["Residual"],
    s=70,
    alpha=0.75
)

# Reference line
ax.axhline(
    0,
    color="red",
    linestyle="--"
)

# Label the three most influential countries
top_influential = diagnostics.nlargest(
    3, "Cooks_D"
)

for _, row in top_influential.iterrows():
    ax.annotate(
        row["Country"],
        (row["Predicted"], row["Residual"]),
        xytext=(5, 5),
        textcoords="offset points"
    )

ax.set_title("OLS Regression Diagnostics — Residuals")
ax.set_xlabel("Predicted change in defence burden (pp)")
ax.set_ylabel("Residual (percentage points)")

ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```


### Regression Diagnostics: Residuals and Influential Observations

Regression diagnostics were conducted to evaluate the stability and suitability of the estimated relationship between geographical distance and changes in military expenditure burden.

Cook's Distance identified four potentially influential observations: Spain (0.436), Portugal (0.416), Latvia (0.202), and Poland (0.185).

Spain and Portugal are located relatively far from Russia but experienced larger increases in military expenditure burden than predicted by the linear model.

Conversely, Latvia and Poland are geographically close to Russia and recorded particularly large increases that also exceeded the model's predictions.

These observations suggest that geographical proximity alone does not fully capture the determinants of European defence expenditure changes.

The residual plot also indicates potential limitations of a simple linear specification, although there is insufficient evidence at this stage to establish a nonlinear relationship.

The next step is to evaluate the sensitivity of the estimated coefficient to influential observations.

Importantly, influential observations are not necessarily erroneous and should not be excluded without methodological justification.


```python

# Define sensitivity scenarios
scenarios = {
    "All EU-27": [],
    "Without Spain": ["Spain"],
    "Without Portugal": ["Portugal"],
    "Without Spain and Portugal": ["Spain", "Portugal"]
}

regression_results = []

for scenario, excluded in scenarios.items():

    # Select countries for the regression
    subset = reg_data[
        ~reg_data["Country"].isin(excluded)
    ].copy()

    # Estimate the same OLS regression
    sensitivity_model = smf.ols(
        formula="Change_pp ~ Distance_100km",
        data=subset
    ).fit(cov_type="HC3")

    # Store the results
    regression_results.append({
        "Scenario": scenario,
        "Countries": len(subset),
        "Coefficient": sensitivity_model.params["Distance_100km"],
        "p-value": sensitivity_model.pvalues["Distance_100km"],
        "R-squared": sensitivity_model.rsquared
    })

# Create comparison table
sensitivity_results = pd.DataFrame(regression_results)

display(sensitivity_results.round(4))
```

```python

loo_results = []

# Estimate a regression excluding one country at a time
for country in reg_data["Country"]:

    subset = reg_data[
        reg_data["Country"] != country
    ]

    loo_model = smf.ols(
        formula="Change_pp ~ Distance_100km",
        data=subset
    ).fit(cov_type="HC3")

    loo_results.append({
        "Excluded_Country": country,
        "Coefficient": loo_model.params["Distance_100km"],
        "R-squared": loo_model.rsquared
    })

loo_results = pd.DataFrame(loo_results)

# Compare with our original regression coefficient
original_coef = model.params["Distance_100km"]

loo_results["Coefficient_Change"] = (
    loo_results["Coefficient"] - original_coef
)

display(
    loo_results.reindex(
        loo_results["Coefficient_Change"].abs()
        .sort_values(ascending=False).index
    ).head(10).round(4)
)
```

```python

fig, ax = plt.subplots(figsize=(10, 6))

# Plot leave-one-out coefficients
ax.scatter(
    range(len(loo_results)),
    loo_results["Coefficient"],
    s=60,
    alpha=0.75
)

# Original regression coefficient
ax.axhline(
    original_coef,
    color="red",
    linestyle="--",
    label=f"Original coefficient: {original_coef:.4f}"
)

ax.set_title(
    "Sensitivity of the Distance Coefficient to Individual Countries"
)

ax.set_xlabel("Country excluded from regression")
ax.set_ylabel("Distance coefficient (pp per 100 km)")

ax.set_xticks(range(len(loo_results)))
ax.set_xticklabels(
    loo_results["Excluded_Country"],
    rotation=90
)

ax.legend()
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

A sensitivity analysis was conducted to assess the stability of the estimated relationship between capital-city distance to Russia and changes in military expenditure burden.

The original OLS regression produced a distance coefficient of **−0.0762 percentage points per 100 km**, with an R-squared of 0.466.

Excluding Spain increased the magnitude of the coefficient to −0.0898, while excluding Portugal produced a coefficient of −0.0901.

When both countries were excluded, the coefficient became −0.1176, with an R-squared of 0.625.

These findings suggest that Spain and Portugal moderate the estimated negative relationship, since both experienced larger increases in military expenditure burden than predicted by their geographical distance.

A leave-one-out sensitivity analysis further demonstrated that the estimated coefficient remained negative when each of the 27 EU member states was excluded individually.

**Overall, the negative association is robust in direction, although its estimated magnitude is sensitive to certain influential observations.**

The full EU-27 sample is retained as the primary specification. Results excluding individual countries are interpreted as sensitivity checks rather than alternative definitive models.

The analysis remains exploratory and does not establish that geographical proximity causes changes in military expenditure.

# 4. Part C — Are European Countries Dependent on Foreign Military Suppliers?

## C1. Research Objective, Data and Methodology

### Research Question

> To what extent do EU member states rely on the United States and other foreign countries for imported major military equipment, and how has this changed over time?

Parts A and B examined the evolution and geographical distribution of European military expenditure.

However, increasing military expenditure does not necessarily strengthen Europe's defence-industrial autonomy. Countries may expand their military capabilities by purchasing equipment from domestic producers, other EU member states, or suppliers outside the European Union.

This section examines the geographical origins of major conventional weapons imported by the EU-27.

### Data Source and Temporal Coverage

The analysis uses the **SIPRI Arms Transfers Database**, which records international transfers of major conventional weapons.

The dataset contains bilateral supplier–recipient relationships for all 27 EU member states, covering two five-year periods:

- 2016–2020
- 2021–2025

These periods allow comparisons of arms-delivery patterns before and after the beginning of the recent acceleration in European rearmament.

The second period includes deliveries that may originate from procurement decisions made before Russia's full-scale invasion of Ukraine in 2022.

### Understanding Trend Indicator Value (TIV)

The Stockholm International Peace Research Institute (SIPRI) uses the **Trend Indicator Value (TIV)** to measure the volume of international transfers of major conventional weapons.

TIV assigns standardized values to military equipment, such as combat aircraft, tanks, missiles, and warships, based on a methodology involving estimated production costs and military characteristics.

This makes it possible to compare transfers involving different weapon systems, countries, and periods.

**TIV is not a monetary value.**

For example, 1,000 million TIV does not mean that a country spent $1 billion purchasing weapons.

TIV is intended to measure the volume of delivered major military equipment rather than actual procurement expenditure.

As an illustration, Poland received approximately 2,453 million TIV from the United States during 2021–2025, representing around 44% of its recorded major arms import volume.

This does not imply that 44% of Poland's entire military inventory or procurement budget was American.

### Data Preparation

The SIPRI export contains separate supplier tables for individual EU recipients.

These tables are transformed into a structured dataset containing four principal variables:

- **Recipient:** EU member state receiving military equipment.
- **Supplier:** Country recorded as supplying the equipment.
- **Period:** Five-year observation window.
- **TIV_million:** Volume of delivered major arms in millions of SIPRI TIV.

The resulting dataset allows analysis of supplier shares, bilateral relationships, concentration, and changes over time.

SIPRI's published recipient totals are retained separately to validate the extraction and calculate import shares.

Blank entries represent no identified deliveries, while a displayed zero can represent a positive transfer below 0.5 million TIV. Published values are rounded, so small reconciliation differences are expected.

### Analytical Scope

The analysis investigates:

1. The composition of EU arms imports by supplier origin.
2. The role of the United States in national arms imports.
3. Geographical differences in US-origin import shares.
4. Supplier concentration and diversification.
5. The structure of supplier–recipient arms-transfer networks.

These indicators measure **reliance on foreign suppliers within recorded major arms imports**, not dependence across countries' entire military inventories or defence industries.

## C4. Supplier Concentration and Diversification

### Research Objective

Dependence on foreign arms suppliers is not determined exclusively by the role of the United States.

A country may obtain most of its imported military equipment from a single supplier or maintain a diversified portfolio of international suppliers.

To investigate this distinction, supplier concentration is measured using the **Herfindahl–Hirschman Index (HHI)**.

### Methodology

For each recipient and period, the share of imported major arms originating from each supplier is calculated.

The HHI is obtained by summing the squared supplier shares:

**HHI = Sum of squared individual supplier shares**

The index approaches 1 when imports are concentrated in a single supplier and decreases as transfers become more evenly distributed across multiple suppliers.

Three complementary indicators are considered:

- **HHI:** Overall concentration of major arms imports.
- **Number of suppliers:** Number of suppliers with positive reported TIV.
- **Effective number of suppliers:** Reciprocal of HHI, representing the equivalent number of equally important suppliers.

The analysis compares these indicators between 2016–2020 and 2021–2025.

A declining HHI indicates greater diversification of the observed import portfolio, whereas an increasing HHI indicates greater concentration.

However, concentration within international imports does not capture dependence on domestic production or the wider defence-industrial supply chain.


```python

# Aggregate TIV by recipient, period and supplier
supplier_data = (
    arms_bilateral.groupby(
        ["Recipient", "Period", "Supplier"],
        as_index=False
    )["TIV_million"]
    .sum()
)

# Keep positive reported TIV values
supplier_data = supplier_data[
    supplier_data["TIV_million"] > 0
].copy()

# Calculate supplier totals for each country-period
supplier_data["Country_Total"] = (
    supplier_data.groupby(
        ["Recipient", "Period"]
    )["TIV_million"]
    .transform("sum")
)

# Calculate each supplier's share
supplier_data["Supplier_Share"] = (
    supplier_data["TIV_million"]
    / supplier_data["Country_Total"]
)

# Square shares for HHI
supplier_data["Share_Squared"] = (
    supplier_data["Supplier_Share"] ** 2
)

display(supplier_data.head(10))
```

```python

# Calculate HHI and additional indicators
supplier_concentration = (
    supplier_data.groupby(["Recipient", "Period"])
    .agg(
        HHI=("Share_Squared", "sum"),
        Number_of_Suppliers=("Supplier", "nunique"),
        Largest_Supplier_Share=("Supplier_Share", "max")
    )
    .reset_index()
)

# Convert largest supplier share to percentage
supplier_concentration["Largest_Supplier_Share"] *= 100

# Calculate effective number of suppliers
supplier_concentration["Effective_Suppliers"] = (
    1 / supplier_concentration["HHI"]
)

# Identify the largest supplier for each country
top_supplier_idx = (
    supplier_data.groupby(["Recipient", "Period"])
    ["Supplier_Share"].idxmax()
)

top_suppliers = supplier_data.loc[
    top_supplier_idx,
    ["Recipient", "Period", "Supplier"]
].rename(columns={"Supplier": "Largest_Supplier"})

supplier_concentration = supplier_concentration.merge(
    top_suppliers,
    on=["Recipient", "Period"],
    validate="one_to_one"
)

display(supplier_concentration.head(10).round(3))
```

```python

# Reshape concentration data
hhi_comparison = supplier_concentration.pivot(
    index="Recipient",
    columns="Period",
    values="HHI"
)

# Rename columns
hhi_comparison = hhi_comparison.rename(
    columns={
        "2016-2020": "HHI_2016_2020",
        "2021-2025": "HHI_2021_2025"
    }
)

# Calculate HHI change
hhi_comparison["HHI_Change"] = (
    hhi_comparison["HHI_2021_2025"]
    - hhi_comparison["HHI_2016_2020"]
)

# Rank by recent concentration
display(
    hhi_comparison.sort_values(
        "HHI_2021_2025",
        ascending=False
    ).round(3)
)
```

```python

# Sort countries by HHI change
hhi_change = hhi_comparison.sort_values(
    "HHI_Change",
    ascending=True
)

fig, ax = plt.subplots(figsize=(11, 10))

ax.barh(
    hhi_change.index,
    hhi_change["HHI_Change"]
)

ax.axvline(
    0,
    color="black",
    linewidth=0.8
)

ax.set_title(
    "Change in Major Arms Import Supplier Concentration",
    fontsize=14
)

ax.set_xlabel(
    "Change in HHI (2021–2025 minus 2016–2020)"
)

ax.set_ylabel("EU Member State")
ax.grid(axis="x", alpha=0.3)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

## C5. European Arms-Transfer Networks

### Research Objective

Previous analyses examined the origins of EU major arms imports, dependence on US suppliers, and supplier concentration.

This section investigates the structure of the bilateral arms-transfer relationships connecting international suppliers with EU member states.

The objective is to identify the largest supplier–recipient relationships, examine the geographical reach of major suppliers, and investigate how these relationships changed between 2016–2020 and 2021–2025.

### Methodology

International arms transfers are represented as a directed, weighted network.

- **Nodes:** Supplier countries and EU recipient countries.
- **Connections:** Recorded major arms transfers from suppliers to recipients.
- **Weights:** SIPRI Trend Indicator Values (TIV), representing the volume of delivered major weapons.

Sankey diagrams are used to visualize the transfer flows.

To maintain readability, the eight largest suppliers and twelve largest recipients across both periods are displayed individually, while the remaining countries are grouped into residual categories.

```python

def create_arms_sankey(period):

    # Select transfers for the requested period
    data = arms_network[
        arms_network["Period"] == period
    ].copy()

    # Group smaller suppliers
    data["Source"] = np.where(
        data["Supplier"].isin(top_suppliers),
        data["Supplier"],
        "Other suppliers"
    )

    # Group smaller EU recipients
    data["Target"] = np.where(
        data["Recipient"].isin(top_recipients),
        data["Recipient"],
        "Other EU recipients"
    )

    # Aggregate grouped relationships
    flows = (
        data.groupby(
            ["Source", "Target"],
            as_index=False
        )["TIV_million"]
        .sum()
    )

    # Create separate supplier and recipient nodes
    sources = flows["Source"].unique().tolist()
    targets = flows["Target"].unique().tolist()

    source_ids = {
        name: i for i, name in enumerate(sources)
    }

    target_ids = {
        name: i + len(sources)
        for i, name in enumerate(targets)
    }

    labels = (
        [f"{s} (supplier)" for s in sources] +
        [f"{r} (recipient)" for r in targets]
    )

    # Create the Sankey diagram
    fig = go.Figure(
        data=[go.Sankey(
            node=dict(
                pad=15,
                thickness=15,
                label=labels
            ),
            link=dict(
                source=flows["Source"].map(source_ids),
                target=flows["Target"].map(target_ids),
                value=flows["TIV_million"]
            )
        )]
    )

    fig.update_layout(
        title_text=f"EU-27 Major Arms Imports — {period}",
        font_size=11,
        height=850
    )

    fig.show()
```

# 5. Part D — Connecting Rearmament, Geography and Arms Imports

## Research Objective

The previous sections examined three dimensions of European defence policy:

1. Changes in military expenditure across EU member states.
2. Geographical differences in defence spending.
3. The origins and concentration of major arms imports.

This section combines these dimensions to investigate whether their observed patterns are statistically related.

Particular attention is given to two hypotheses:

**H2:** Countries experiencing greater growth in real military expenditure also tend to experience greater growth in imported major arms.

**H3:** Countries geographically closer to Russia obtain a higher proportion of their imported major arms from the United States.

### Data Integration

A common country-level dataset is constructed for the current EU-27.

Military expenditure is aggregated over the two five-year periods used in Part C: 2016–2020 and 2021–2025.

For each country, the dataset contains:

- Real military expenditure growth between the two periods.
- Growth in imported major arms TIV.
- US-origin major arms import shares.
- Territorial distance from Russia.
- Capital-city distance from Russia.

Each EU member state represents one observation in the combined dataset.

### Methodological Considerations

Military expenditure and major arms imports are measured in different units and represent different defence activities.

Consequently, the analysis compares patterns and statistical associations rather than equating financial expenditure with delivered weapons.

Additionally, the timing of military procurement decisions and equipment deliveries may differ substantially.

The following analyses are therefore exploratory and cannot establish direct causal relationships between geography, expenditure, and arms imports.


```python

# Select the years covered by our arms-transfer analysis
spending_periods = eu_defence[
    eu_defence["Year"].between(2016, 2025)
].copy()

# Assign each year to its corresponding period
spending_periods["Period"] = np.where(
    spending_periods["Year"] <= 2020,
    "2016-2020",
    "2021-2025"
)

# Sum military expenditure across each five-year period
spending_periods = (
    spending_periods.groupby(
        ["Country", "Period"],
        as_index=False
    )["Military_Expenditure"]
    .sum()
)

# Reshape the dataset
spending_comparison = spending_periods.pivot(
    index="Country",
    columns="Period",
    values="Military_Expenditure"
).rename(columns={
    "2016-2020": "Spending_2016_2020",
    "2021-2025": "Spending_2021_2025"
})

# Calculate percentage change
spending_comparison["Spending_Growth_Percent"] = (
    (
        spending_comparison["Spending_2021_2025"]
        / spending_comparison["Spending_2016_2020"]
    ) - 1
) * 100

display(spending_comparison.round(2))
```

To examine this relationship, percentage changes in aggregate real military expenditure are compared with percentage changes in major arms import TIV between 2016–2020 and 2021–2025.

Spearman's rank correlation is used to evaluate whether countries with greater spending increases also tend to rank higher in arms-import growth.

This method reduces sensitivity to extreme numerical values, which may arise when a country records very small import volumes during the initial period.

Countries without positive recorded imports in the first period cannot be assigned a conventional percentage import-growth rate and are excluded from this correlation.

Several factors may explain why increases in military expenditure do not correspond directly to increases in arms-delivery volumes.

Additionally, procurement decisions and equipment deliveries may be separated by several years.

The results therefore suggest that greater military expenditure cannot be assumed to produce an immediate or proportionate increase in major arms imports.

## D3 — Multivariate Regression: Geography, Military Spending and US Arms Imports

### Research Objective

This section investigates whether geographical proximity to Russia is associated with the share of major arms imported from the United States, after accounting for differences in military expenditure growth across EU member states.

### Methodology

Two Ordinary Least Squares (OLS) regression models are estimated.

**Model 1** examines the relationship between geographical distance from Russia and the US share of major arms imports.

**Model 2** introduces military expenditure growth as an additional explanatory variable.

The dependent variable is the percentage of major arms imported from the United States during 2021–2025, based on SIPRI Trend Indicator Values (TIV).

The explanatory variables are:

1. **Capital distance to Russia:** Measured in kilometres, using geographical data prepared in Part B.
2. **Military expenditure growth:** Percentage change in aggregate real military expenditure between 2016–2020 and 2021–2025.

HC3 heteroskedasticity-robust standard errors are used.

### Analytical Purpose

By comparing the two models, the analysis examines whether the relationship between geographical proximity and US-origin arms imports persists after introducing military expenditure growth.

The models are intended to explore statistical associations rather than establish causal relationships.

### Limitations

The analysis is based on a small cross-sectional sample of EU member states.

US import shares are bounded between 0% and 100%, making linear regression an approximation.

Military procurement decisions are influenced by multiple factors not included in the model, including domestic defence production, historical procurement relationships, alliance commitments, industrial policies, and the timing of equipment deliveries.

The dependent and explanatory variables also cover overlapping periods, so the models cannot establish temporal causality.

Consequently, the results should be interpreted as exploratory evidence about the relationship between geography, military expenditure, and supplier composition.


```python

import statsmodels.formula.api as smf

# Select variables for regression
reg_data = combined_analysis[
    [
        "Country",
        "Capital_Distance_Russia_km",
        "Spending_Growth_Percent",
        "US_Share_2021_2025"
    ]
].dropna().copy()

# Convert distance to units of 100 km
reg_data["Distance_100km"] = (
    reg_data["Capital_Distance_Russia_km"] / 100
)

# Convert spending growth to units of 10 percentage points
reg_data["Spending_Growth_10pp"] = (
    reg_data["Spending_Growth_Percent"] / 10
)

print("Countries included:", len(reg_data))
display(reg_data.head().round(2))
```

```python

# Model 1: Geography only
model_d1 = smf.ols(
    formula="US_Share_2021_2025 ~ Distance_100km",
    data=reg_data
).fit(cov_type="HC3")

# Model 2: Geography + spending growth
model_d2 = smf.ols(
    formula=(
        "US_Share_2021_2025 ~ "
        "Distance_100km + Spending_Growth_10pp"
    ),
    data=reg_data
).fit(cov_type="HC3")

print("MODEL 1 — Geography only")
print(model_d1.summary())

print("\nMODEL 2 — Geography + Spending Growth")
print(model_d2.summary())
```

## D4 — Robustness and Sensitivity Analysis

### Research Objective

This section evaluates the robustness of the relationship between geographical proximity to Russia and the share of EU member states' major arms imports originating from the United States.

### Methodology

Four complementary robustness checks are conducted.

**1. Alternative geographical measurement**

The regression is re-estimated using the shortest distance between national territory and Russia, rather than the distance from each country's capital.

This examines whether the estimated relationship depends on how geographical proximity is measured.

**2. Leave-one-out sensitivity**

The main model is re-estimated repeatedly, excluding one EU member state at a time.

This helps identify whether individual countries disproportionately influence the estimated relationship.

**3. Import-volume-weighted regression**

A Weighted Least Squares regression is estimated using each country's total major arms import TIV during 2021–2025 as its weight.

This examines whether the relationship changes when countries importing larger volumes of military equipment receive greater emphasis.

**4. Historical US import shares**

A further regression includes the US share of major arms imports during 2016–2020 as an explanatory variable.

This investigates whether geographical proximity provides additional explanatory information beyond previous supplier relationships.

### Interpretation

The robustness checks are evaluated by comparing the direction, magnitude, uncertainty, and stability of the geographical distance coefficient across model specifications.

Consistent coefficients strengthen confidence in the observed statistical association, while substantial variation suggests that the relationship is sensitive to modelling choices or influential observations.

### Limitations

The analysis relies on a small cross-sectional sample of EU member states.

The regressions describe associations rather than causal effects.

Historical arms-import shares are imperfect indicators of established defence relationships, while import-volume weighting changes the interpretation of the estimated relationship.

Moreover, military expenditure growth and arms-import shares are measured over overlapping periods, limiting causal and temporal interpretations.


```python

# Create a common sample for both models
robust_data = combined_analysis[
    [
        "Country",
        "Distance_Russia_km",
        "Capital_Distance_Russia_km",
        "Spending_Growth_Percent",
        "US_Share_2021_2025"
    ]
].dropna().copy()

# Express distances per 100 km
robust_data["Territorial_Distance_100km"] = (
    robust_data["Distance_Russia_km"] / 100
)

robust_data["Capital_Distance_100km"] = (
    robust_data["Capital_Distance_Russia_km"] / 100
)

# Express spending growth per 10 percentage points
robust_data["Spending_Growth_10pp"] = (
    robust_data["Spending_Growth_Percent"] / 10
)

# Model using capital distance
capital_model = smf.ols(
    "US_Share_2021_2025 ~ "
    "Capital_Distance_100km + Spending_Growth_10pp",
    data=robust_data
).fit(cov_type="HC3")

# Model using territorial distance
territorial_model = smf.ols(
    "US_Share_2021_2025 ~ "
    "Territorial_Distance_100km + Spending_Growth_10pp",
    data=robust_data
).fit(cov_type="HC3")

# Compare distance coefficients
distance_comparison = pd.DataFrame({
    "Distance measure": [
        "Capital distance",
        "Territorial distance"
    ],
    "Coefficient": [
        capital_model.params["Capital_Distance_100km"],
        territorial_model.params["Territorial_Distance_100km"]
    ],
    "P_value": [
        capital_model.pvalues["Capital_Distance_100km"],
        territorial_model.pvalues["Territorial_Distance_100km"]
    ],
    "R_squared": [
        capital_model.rsquared,
        territorial_model.rsquared
    ]
})

display(distance_comparison.round(4))
```

```python

# Store results from each regression
loo_results = []

for country in robust_data["Country"]:

    # Exclude one country
    subset = robust_data[
        robust_data["Country"] != country
    ]

    # Re-estimate the model
    loo_model = smf.ols(
        "US_Share_2021_2025 ~ "
        "Capital_Distance_100km + Spending_Growth_10pp",
        data=subset
    ).fit(cov_type="HC3")

    loo_results.append({
        "Excluded_Country": country,
        "Distance_Coefficient": loo_model.params[
            "Capital_Distance_100km"
        ],
        "Distance_p_value": loo_model.pvalues[
            "Capital_Distance_100km"
        ],
        "R_squared": loo_model.rsquared
    })

# Create comparison DataFrame
loo_results = pd.DataFrame(loo_results)

display(
    loo_results.sort_values(
        "Distance_Coefficient"
    ).round(4)
)
```

```python

# Extract recent-period import totals
recent_import_totals = arms_totals[
    arms_totals["Period"] == "2021-2025"
][["Recipient", "Total_TIV_million"]].rename(
    columns={"Recipient": "Country"}
)

# Add import totals to regression data
weighted_data = robust_data.merge(
    recent_import_totals,
    on="Country",
    validate="one_to_one"
)

# Keep countries with positive import volumes
weighted_data = weighted_data[
    weighted_data["Total_TIV_million"] > 0
].copy()

# Estimate weighted regression
weighted_model = smf.wls(
    "US_Share_2021_2025 ~ "
    "Capital_Distance_100km + Spending_Growth_10pp",
    data=weighted_data,
    weights=weighted_data["Total_TIV_million"]
).fit(cov_type="HC3")

print(weighted_model.summary())
```

## D5. Summary of Hypothesis Testing

The final stage of the combined analysis consolidates the statistical results from Parts B and D.

Three hypotheses are evaluated:

- **H1:** Geographical proximity to Russia is associated with greater increases in the military expenditure burden.
- **H2:** Real military expenditure growth is positively associated with growth in major arms imports.
- **H3:** Geographical proximity to Russia is associated with higher US-origin major arms import shares.

Spearman correlations provide comparable indicators of the direction and strength of these relationships.

The multivariate regression models and sensitivity checks provide additional evidence about H3.

The hypotheses are evaluated using the magnitude, direction, statistical uncertainty, and robustness of the estimated associations.

Statistical significance is not interpreted as proof of causality, and non-significant results are not treated as evidence that the corresponding relationships are impossible or absent.


```python

from scipy.stats import spearmanr

def summarize_correlation(data, x, y, label):
    subset = data[[x, y]].dropna()

    rho, p = spearmanr(
        subset[x],
        subset[y]
    )

    return {
        "Relationship": label,
        "Spearman_rho": rho,
        "P_value": p,
        "N": len(subset)
    }


final_correlations = pd.DataFrame([
    summarize_correlation(
        geo_analysis,
        "Capital_Distance_Russia_km",
        "Change_pp",
        "Distance vs GDP defence burden change"
    ),

    summarize_correlation(
        combined_analysis,
        "Spending_Growth_Percent",
        "Import_Growth_Percent",
        "Spending growth vs arms-import growth"
    ),

    summarize_correlation(
        combined_analysis,
        "Capital_Distance_Russia_km",
        "US_Share_2021_2025",
        "Distance vs US arms import share"
    )
])

display(final_correlations.round(4))
```

```python

hypothesis_summary = final_correlations.copy()

hypothesis_summary["Hypothesis"] = [
    "H1: Closer countries increase GDP defence burden more",
    "H2: Spending growth relates positively to import growth",
    "H3: Closer countries have higher US import shares"
]

# Expected sign: H1 negative, H2 positive, H3 negative
expected_signs = [-1, 1, -1]

hypothesis_summary["Expected_Direction"] = expected_signs

hypothesis_summary["Expected_Sign"] = [
    (rho * direction > 0)
    for rho, direction in zip(
        hypothesis_summary["Spearman_rho"],
        expected_signs
    )
]

hypothesis_summary["P_below_0_05"] = (
    hypothesis_summary["P_value"] < 0.05
)

display(
    hypothesis_summary[
        [
            "Hypothesis",
            "Spearman_rho",
            "P_value",
            "N",
            "Expected_Sign",
            "P_below_0_05"
        ]
    ].round(4)
)
```

# 6.Discussion and Conclusions

## E1. Is Europe Really Rearming?

The analysis provides strong descriptive evidence that EU member states have substantially increased their military expenditure.

Between 2014 and 2025, aggregate EU-27 military expenditure increased from approximately $210.8 billion to $430.4 billion, expressed in constant 2024 US dollars.

This represents a real increase of approximately 104%.

Furthermore, annual expenditure growth accelerated after 2022, indicating a significant change in the pace of European rearmament.

However, this increase is not uniform across countries.

Some member states, particularly those in Central, Eastern, and Northern Europe, have experienced particularly strong increases in military expenditure and defence spending relative to GDP.

Overall, the findings support the conclusion that Europe is undergoing a substantial rearmament process, although the intensity and characteristics of this process differ across member states.

## E2. Does Geography Matter?

The geographical analysis reveals a strong negative association between proximity to Russia and the share of GDP devoted to military expenditure.

Countries geographically closer to Russia generally exhibit higher defence burdens and larger increases in those burdens over the period examined.

This relationship remains visible when considering alternative geographical measures and several robustness checks.

The results are consistent with the interpretation that geographical exposure to Russia is associated with stronger defence spending responses.

However, proximity should not be interpreted as a direct causal mechanism.

Defence expenditure is influenced by several additional factors, including threat perceptions, NATO commitments, national security strategies, historical experiences, fiscal capacity, and political priorities.

Geography therefore provides an informative explanatory variable, but not a complete explanation of European rearmament.

## E3. Are European Countries Dependent on American Weapons?

The arms-transfer analysis reveals that the United States remains the largest individual foreign supplier of major military equipment to EU member states.

During 2021–2025, approximately half of the recorded volume of EU major arms imports originated from the United States, based on SIPRI Trend Indicator Values (TIV).

However, reliance on American suppliers differs substantially across member states.

Countries such as Italy, Slovakia, the Netherlands, Sweden, and Denmark received particularly high proportions of their imported major military equipment from the United States during the recent period.

Other countries displayed different procurement patterns. Poland, for example, received substantial volumes of major military equipment from South Korea as well as the United States.

The findings also show the increasing importance of other non-EU suppliers, including South Korea, Israel, Norway, and the United Kingdom.

Consequently, European arms imports are characterized by substantial reliance on suppliers outside the European Union, although the composition of this reliance varies considerably between countries.


## E4. How Are Rearmament, Geography, and Arms Imports Connected?

The combined analysis reveals an important distinction between the geographical distribution of military expenditure and the origins of imported military equipment.

The strongest statistical relationship identified in this study concerns geographical proximity to Russia and changes in military expenditure relative to GDP.

Countries closer to Russia generally experienced larger increases in their defence expenditure burden, consistent with the interpretation that geopolitical exposure is associated with national defence spending priorities.

However, the relationship between military expenditure growth and the growth of imported major arms is considerably weaker.

The observed correlation is positive but not statistically significant, suggesting that increasing military expenditure does not necessarily translate directly into greater volumes of delivered imported weapons.

Similarly, the analysis does not identify a strong or statistically significant relationship between geographical distance from Russia and US-origin major arms import shares.

The multivariate regression and sensitivity analyses reinforce this conclusion, showing that the estimated geographical relationship is weak and unstable across alternative specifications.

These findings suggest that **the factors associated with increased defence expenditure are not necessarily the same as those shaping international military procurement**.

A country may respond to perceived geopolitical threats by increasing defence expenditure while allocating those resources to personnel, domestic arms production, European suppliers, or non-European suppliers.

Moreover, military procurement decisions may reflect established industrial relationships, technological requirements, equipment interoperability, and delivery schedules.

The results therefore support interpreting European rearmament and foreign supplier reliance as related but analytically distinct processes.

The study provides evidence of a geographical pattern in defence spending, but not convincing evidence that this pattern translates directly into a geographical pattern of American arms-import reliance.


## E5. Implications for European Strategic Autonomy

The findings highlight a potential tension between European rearmament and European defence-industrial autonomy.

Increasing military expenditure strengthens the resources available for defence, but does not automatically increase the capacity of European industries to develop and produce military equipment independently.

If additional military resources are directed toward major weapons manufactured outside the EU, European armed forces may become better equipped while maintaining or expanding their reliance on foreign suppliers.

However, imports alone do not provide a complete measure of strategic autonomy.

Operational independence also depends on access to maintenance, spare parts, software, weapons integration, industrial know-how, production capacity, and critical technologies.

Therefore, the arms-import analysis should be interpreted as one indicator of European defence-industrial reliance rather than a comprehensive measure of strategic autonomy.



## E6. Research Limitations

This study has several methodological limitations.

### 1. Observational analysis

The statistical relationships identified in the study are observational and do not establish causality.

In particular, the association between geographical proximity to Russia and military expenditure cannot by itself demonstrate that proximity caused changes in defence policy.

### 2. Small sample size

The analysis focuses on the 27 EU member states.

This limits statistical power, particularly in multivariate regressions, and makes some estimates sensitive to individual countries.

### 3. Different measurement concepts

Military expenditure and arms transfers measure different aspects of defence activity.

Military expenditure is expressed in monetary values, while SIPRI TIV measures the volume of major conventional weapons delivered internationally.

The two indicators should not be interpreted as directly interchangeable.

### 4. Procurement and delivery delays

Major military equipment can take several years to produce and deliver.

Consequently, arms transfers recorded during 2021–2025 may reflect contracts and procurement decisions made in earlier years.

### 5. Incomplete representation of defence-industrial autonomy

The analysis does not include domestic military production, existing equipment inventories, technological capabilities, maintenance dependencies, or the full industrial supply chains of multinational weapons systems.

### 6. Geographical simplification

Distance to Russia provides a consistent geographical indicator, but does not fully capture security perceptions, historical relationships, military geography, or geopolitical exposure.

### 7. Classification and rounding

Supplier categories use a consistent EU-27 membership definition across the study period. They are not historical classifications of EU membership in each year.

SIPRI TIV figures are rounded, and special transfer records, including same-country supplier–recipient entries, require careful treatment.

These limitations do not invalidate the study, but define the scope within which its conclusions can be interpreted.



## E7. Overall Conclusion

This study investigated European rearmament through three interconnected dimensions: the evolution of military expenditure, geographical proximity to Russia, and reliance on foreign suppliers of major conventional weapons.

The findings demonstrate that the EU-27 has substantially increased its military expenditure. Between 2014 and 2025, aggregate real military expenditure more than doubled, with particularly strong increases in several countries geographically closer to Russia.

The geographical analysis identifies a strong negative association between distance from Russia and increases in military expenditure relative to GDP. This suggests that geographical exposure is an important correlate of European rearmament, although the observational analysis cannot establish causality.

At the same time, the arms-transfer analysis reveals substantial reliance on suppliers outside the European Union.

The United States remains the largest individual supplier of imported major military equipment, accounting for approximately half of the EU-27's recorded major arms import volume during 2021–2025. Other non-EU suppliers have also gained importance.

However, the combined statistical analysis does not establish that countries increasing military expenditure more strongly also experience systematically greater growth in major arms imports.

Nor does it provide convincing evidence that countries geographically closer to Russia import a systematically higher proportion of their major weapons from the United States.

These findings highlight a central distinction:

**Geography appears strongly associated with how European countries allocate resources to defence, but much less clearly with where they obtain their imported military equipment.**

This distinction has important implications for European strategic autonomy.

Increasing defence expenditure may strengthen European military capabilities, but greater spending does not automatically produce a more independent defence-industrial base.

European strategic autonomy depends not only on the quantity of military resources available, but also on industrial production capacity, technological capabilities, procurement decisions, maintenance systems, and the structure of international defence supply chains.

Ultimately, Europe's rearmament should be assessed along two related but distinct dimensions: the expansion of military capabilities and the development of an autonomous and resilient European defence-industrial capacity.

The study provides evidence of substantial European rearmament, while demonstrating that the geographical and industrial dimensions of this process follow different empirical patterns.
